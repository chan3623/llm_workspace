## raw SDK- openai


```python
# basic.py
from common.common import OPENAI_MODEL, get_openai_client

# 1. raw SDK 클라이언트 생성 (.env에서 OPENAI_API_KEY 자동 검증)
client = get_openai_client()

# 2. OpenAI Chat Completions API 호출
response = client.chat.completions.create(
    model=OPENAI_MODEL,
    temperature=0,
    messages=[{"role": "user", "content": "인공지능에 대해 설명해줘"}],
)

# 3. 응답 텍스트 출력
print(response.choices[0].message.content)
```

## raw SDK - gemini


```python
from common.common import GEMINI_MODEL, get_genai_client

# 1. google-genai Client 생성
client = get_genai_client()

# 2. Gemini 모델 호출 (google-genai SDK 전용 메서드)
response = client.models.generate_content(
    model=GEMINI_MODEL,
    contents="인공지능에 대해 설명해줘",
)

# 3. 응답 텍스트 출력
print(response.text)
```

## langchain - openai


```python
# basic.py
from common.common import get_chat

# 1. LangChain ChatOpenAI 객체 생성 (provider="openai" 전달)
gpt = get_chat(provider="openai", temperature=0)

# 2. LangChain 방식으로 호출
response = gpt.invoke("인공지능에 대해 설명해줘")

# 3. 결과 출력 (LangChain AIMessage 객체이므로 .content 사용 가능)
print(response.content)
```
