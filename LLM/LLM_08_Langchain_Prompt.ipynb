✅ PromptTemplate (from langchain.prompts)

- 텍스트 기반 LLM (예: GPT-3.5, GPT-4 텍스트모드)용
- 하나의 문자열을 생성
- 단순하고 직관적인 형식

✅ ChatPromptTemplate (from langchain_core.prompts)

- 챗 기반 LLM (예: ChatGPT, GPT-4o)용
- 역할(role: system/user/assistant 등)을 기반으로 구성된 메시지 리스트 사용
- 대화형 프롬프트를 정의할 수 있음

### PromptTemplate


```python
template = "{product} 제품 홍보문구 작성해줘"
prompt = PromptTemplate(input_variables=["product"], template=template)
txt = prompt.format(product="컴퓨터")
txt
```

```python
template = "{product} 제품 홍보문구 {desc}"
prompt = PromptTemplate(input_variables=["product", "desc"], template=template)
txt = prompt.format(product="컴퓨터", desc="설명해줘")
txt
```

```python
template = "오늘은 {day}입니다. {name}님, 좋은 하루 되세요."
prompt = PromptTemplate(
    input_variables=["name"], partial_variables={"day": "화요일"}, template=template
)
txt = prompt.format(name="홍길동")
txt
```

```python
from langchain_core.messages import SystemMessage, HumanMessage, AIMessage
from langchain_openai import ChatOpenAI

template = "{product} 제품 홍보문구 작성해줘"
prompt = PromptTemplate(input_variables=["product"], template=template)

gpt = ChatOpenAI(model="gpt-4o-mini", temperature=0)
rst = gpt.invoke(prompt.format(product="컴퓨터"))
# rst =gpt.invoke( "컴퓨터 제품 홍보문구 작성해줘" )
# rst =gpt.invoke( [HumanMessage(content= prompt.format(product="컴퓨터") ) ] )
print(rst.content)
```

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_openai import ChatOpenAI

prompt = ChatPromptTemplate(
    [
        ("system", "You are a helpful teacher 한국어로 대답해"),
        ("human", "Are ther other measure than time complexity for an {algorithm}"),
        (
            "ai",
            "Yes, there are other measures besides time complexity for an {algorithm}, such as space complexity.",
        ),
        ("human", "What is it?"),
    ]
)

model = ChatOpenAI(model="gpt-4o-mini", temperature=0)

result = model.invoke(prompt.format_messages(algorithm="algorithm"))
result.content
```

```python
import langchain_core.prompts as pt

print(dir(pt))
```

```python
from langchain_core.prompts import FewShotPromptTemplate
```

```python
exmaples = [
    {"input": "Hello", "output": "안녕하세요"},
    {"input": "Goodbye", "output": "안녕히 가세요"},
]

exmaple_promt = PromptTemplate(
    input_variables=["input", "output"], template="Input: {input}\nOutput: {output}"
)

few_shot_prompt = FewShotPromptTemplate(
    examples=exmaples,
    example_prompt=exmaple_promt,
    suffix="Input: {input}\nOutput:",
    input_variables=["input"],
)

fprompt = few_shot_prompt.format(input="Good night")
print(fprompt)
```

```python
gpt = ChatOpenAI(model="gpt-4o-mini", temperature=0)
rst = gpt.invoke(fprompt)
print(rst.content)
```
